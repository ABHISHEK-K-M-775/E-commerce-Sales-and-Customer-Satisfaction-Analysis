```python
import pandas as pd
import os
import logging
import time
from sqlalchemy import create_engine

logging.basicConfig(
    filename='logs.ingestion.db',
    level='INFO',
    format="%(asctime)s-%(levelname)s-%(message)s",
    filemode='a')

engine=create_engine('sqlite:///sales_customer.db')

files=os.listdir('F:\Projects\E-commerce Sales and Customer Satisfaction Analysis\dataset')

def ingest_db(df,table_name,engine):
    df.to_sql(table_name,con=engine,if_exists='replace',index=False)

def load_raw_data():
    start=time.time()    
    for file in files:
        if '.csv' in file:
            df=pd.read_csv('dataset/'+file)
            logging.info(f'ingesting {file} into db')
            ingest_db(df,file[:-4],engine)
    end=time.time()
    logging.info('-------------------ingestion completed-------------------')
    total=(end-start)/60
    logging.info(f'total time for ingestion {total} minutes')   
    
if __name__=='__main__':
    load_raw_data()     

```
